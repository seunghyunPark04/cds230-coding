```python
# load the raisin dataset
df = pd.read_csv('Raisin_Dataset.csv')
df.head()
```

```python
# check data types and missing values
df.info()
```

```python
# convert raisin types into numerical labels
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['Class'] = le.fit_transform(df['Class'])
df.head()
```

```python
# separate input features (x) and target (y)
x = df.drop('Class', axis=1)
y = df['Class']
```

```python
# use 80% for training and 20% for testing
x_train, x_test, y_train, y_test = train_test_split(
    x, y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
```

```python
# standardize features to the same scale
scaler = StandardScaler()

x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
# create and train the svm model using rbf kernel
model = SVC(kernel='rbf', random_state=42)
model.fit(x_train, y_train)
```

```python
# predict raisin types using test data
y_pred = model.predict(x_test)
```

```python
# evaluate the model using accuracy, precision, recall, and f1-score
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
print(classification_report(
    y_test,
    y_pred,
    target_names=le.classes_
))
```

I used the UCI Raisin dataset to predict two types of raisins.

The dataset contains 900 raisins. I used 80% of the data to train the SVM model and 20% to test it.

The SVM model achieved an accuracy of 87.78%, meaning it correctly classified 158 out of 180 raisins in the test dataset.

The results show that SVM can identify different types of raisins based on their physical characteristics, but it can still make some mistakes.
